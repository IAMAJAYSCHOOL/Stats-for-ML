```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("diabetes.csv",dtype={"Outcome":bool})
```

```python
print("shape:",df.shape)
print("\ncolumns:",df.columns.to_list())
print("\nDatatypes:")
print(df.dtypes) 
print("\nDataset Information:")
df.info()
```

```python
print(df.describe())
print("\nVariance:")
print(df.var())

print("\nMedian:")
print(df.median())

print("\nMode:")
print(df.mode().iloc[0])
```

```python
df.hist(figsize=(12,10))
plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4,2, figsize=(6, 6))

for ax, col in zip(axes.flatten(), df.columns[:-1]):
    sns.boxplot(x=df[col], ax=ax)
    #ax.set_title(col)


plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(4,3))
sns.countplot(x="Outcome", data=df)
plt.ylabel("Frequency")
plt.title("Diabetes Outcome")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x="Glucose",
    y="BMI",
    hue="Outcome",
    s=40
)
plt.title("Glucose vs BMI")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(
    data=df,
    x="Age",
    y="Glucose",
    hue="Outcome"
)
plt.title("Age vs Glucose")
plt.show()
```

```python
selected = df[
    [
        "Glucose",
        "BMI",
        "Age",
        "Outcome"
    ]]
sns.pairplot(selected, hue="Outcome",height=1.5,plot_kws={'s': 15})
plt.show()
```

```python
print("Standard Deviations")
print(df.std().sort_values(ascending=False))

print("\nDifference Between Mean and Median")
difference = abs(df.mean() - df.median())
print(difference.sort_values(ascending=False))

print("\nOutcome Frequency")
print(df["Outcome"].value_counts())
```
