# Experiment 4: Statistical Estimation and Hypothesis Testing

**Aim:** To apply statistical estimation and hypothesis testing techniques to the Pima Indians Diabetes Dataset and draw conclusions about population characteristics using sample data.

## 1. Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
from scipy import stats

df = pd.read_csv("diabetes.csv")

print(df.shape)
df.head()
```

## 2. Select Variables and Create Groups

```python
# Separate diabetic and non-diabetic patients

diabetic = df[df["Outcome"] == 1]
non_diabetic = df[df["Outcome"] == 0]

print("Diabetic patients:", len(diabetic))
print("Non-diabetic patients:", len(non_diabetic))
```

## 3. Calculate Point Estimates

```python
# Sample mean of Glucose
mean_glucose = df["Glucose"].mean()

# Sample proportion of diabetic patients
diabetes_proportion = df["Outcome"].mean()

print("Estimated mean Glucose:", mean_glucose)
print("Estimated proportion of diabetic patients:", diabetes_proportion)
```

## 4. Construct a 95% Confidence Interval for Mean Glucose

```python
glucose = df["Glucose"]

n = len(glucose)
mean = glucose.mean()
std_error = stats.sem(glucose)

confidence_interval = stats.t.interval(
    0.95,
    df=n - 1,
    loc=mean,
    scale=std_error
)

print("Sample Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```

## 5. Formulate Hypotheses

```python
# H0: Mean Glucose = 120
# H1: Mean Glucose != 120

reference_value = 120

print("H0: Mean Glucose = 120")
print("H1: Mean Glucose != 120")
```

## 6. Perform One-Sample t-Test

```python
t_stat, p_value = stats.ttest_1samp(
    df["Glucose"],
    popmean=reference_value
)

print("T-statistic:", t_stat)
print("P-value:", p_value)
```

## 7. Make Decision Using Significance Level

```python
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
    print("There is sufficient evidence that the mean Glucose level differs from 120.")
else:
    print("Fail to reject H0")
    print("There is not sufficient evidence that the mean Glucose level differs from 120.")
```

## 8. Perform Two-Sample t-Test for BMI

```python
# Compare BMI of diabetic and non-diabetic patients

t_stat_bmi, p_value_bmi = stats.ttest_ind(
    diabetic["BMI"],
    non_diabetic["BMI"],
    equal_var=False
)

print("T-statistic:", t_stat_bmi)
print("P-value:", p_value_bmi)
```

## 9. Create Categories and Perform Chi-Square Test

```python
# Test association between Age Group and Outcome

df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[0, 30, 50, 100],
    labels=["Young", "Middle", "Older"]
)

contingency_table = pd.crosstab(
    df["AgeGroup"],
    df["Outcome"]
)

print(contingency_table)

chi2, p_chi, dof, expected = stats.chi2_contingency(
    contingency_table
)

print("\nChi-square statistic:", chi2)
print("P-value:", p_chi)
```

## 10. Interpret Chi-Square Test

```python
alpha = 0.05

print("Chi-Square Test")
print("H0: Age Group and Diabetes Outcome are independent.")
print("H1: Age Group and Diabetes Outcome are associated.")

if p_chi < alpha:
    print("\nReject H0")
    print("There is a statistically significant association between Age Group and Diabetes Outcome.")
else:
    print("\nFail to reject H0")
    print("There is no statistically significant evidence of an association between Age Group and Diabetes Outcome.")
```
